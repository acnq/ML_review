$$
\textbf{z} = \textbf{W}^{(1)}\textbf{x} \\
\textbf{h} = ϕ(\textbf{z}) \\ 
\textbf{o} = \textbf{W}^{(2)}\textbf{h} \\ 
L = l(\textbf{o}, y) \\
s = \fracλ2(||\textbf{W}^{(1)}||_F^2 + ||\textbf{W}^{(2)}||_F^2 ) \\ 
J = L + s
$$

### 4.7.3 Backward Propagation

$$
\frac{∂\mathsf{Z}}{∂\mathsf{X}} = \text{prod}(\frac{∂\mathsf{Z}}{∂\mathsf{Y}}, \frac{∂\mathsf{Y}}{∂\mathsf{X}}) \\ 
\frac{∂J}{∂L} = 1 \text{ and } \frac{∂J}{\partial s} = 1 \\ 
\frac{∂J}{∂\textbf{o}} = \text{prod}(\frac{∂J}{∂L}, \frac{∂L}{∂\textbf{o}}) = \frac{∂L}{∂\textbf{o}} \in \mathbb{R}^q \\ 
\frac{∂s}{∂\textbf{W}^{(1)}} = λ\textbf{W}^{(1)} \text{ and } \frac{∂s}{∂\textbf{W}^{(2)}} = λ\textbf{W}^{(2)}  \\ 
\frac{∂J}{∂\textbf{W}^{(2)}} = \text{prod}(\frac{∂J}{∂\textbf{o}}, \frac{∂\textbf{o}}{∂\textbf{W}^{(2)}}) + \text{prod}(\frac{∂J}{∂s}, \frac{∂s}{∂\textbf{W}^{(2)}}) = \frac{∂J}{∂\textbf{o}}\textbf{h}^T + λ\textbf{W}^{(2)} \in \mathbb{R}^{q×h}\\ 
\frac{∂J}{∂\textbf{h}} = \text{prod}(\frac{∂J}{∂\textbf{o}}, \frac{∂\textbf{o}}{∂\textbf{h}}) = {\textbf{W}^{(2)}}^T\frac{∂J}{∂\textbf{o}} \\
\frac{∂J}{∂\textbf{z}} = \text{prod}(\frac{∂J}{∂\textbf{h}}, \frac{∂\textbf{h}}{∂\textbf{z}}) = \frac{∂J}{∂\textbf{h}} ⊙ ϕ'(\textbf{z}) \\
\frac{∂J}{∂\textbf{W}^{(1)}} = \text{prod}(\frac{∂J}{∂\textbf{z}}, \frac{∂\textbf{z}}{∂\textbf{W}^{(1)}}) + \text{prod}(\frac{∂J}{∂s}, \frac{∂s}{∂\textbf{W}^{(1)}}) = \frac{∂J}{∂\textbf{z}}\textbf{x}^T + λ\textbf{W}^{(1)} \in \mathbb{R}^{h×d}\\ 
$$
here $\odot$ means product elementwisely
