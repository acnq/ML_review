### 4.9.3 Distribution Shift Correction

#### 4.9.3.1 Empirical Risk and True Risk
$$
\text{minimize}_f\frac1n\sum_{i=1}^nl(f(\textbf{x}_i, y_i))
$$
true risk define with true distribution $p(\textbf{x}, y)$:
$$
E_{p(\textbf{x}, y)}[l(f(\textbf{x}), y)] = \int\int l(f(\textbf{x}), y)p(\textbf{x}, y)d\textbf{x}dy
$$

#### 4.9.3.2 Covariance Shift Correction
observation value $\textbf{x}_i$ comes from source dist $q(\textbf{x})$, and the target dist is $p(\textbf{x})$, suppose $p(y | \textbf{x}) = q(y | \textbf{x})$

$$
∫∫ l(f(\textbf{x}, y))p(y | \textbf{x})p(\textbf{x}) d\textbf{x}dy = ∫∫l(f(\textbf{x}), y)q(y | \textbf{x}) q(\textbf{x}) \frac{p(\textbf{x})}{q(\textbf{x})}d\textbf{x}dy
$$

new weight:
$$
β_i \stackrel{\text{def}}{=} \frac{p(\textbf{x})}{q(\textbf{x})}
$$
new model:
$$
\text{minimize}_f \frac1n\sum_{i=1}^nβ_il(f(\textbf{x}_i, y_i))
$$
suppose extract samples from both dist, labeled with $z$, if from $p$ then 1, else -1
$$
P(z = 1 | \textbf{x}) = \frac{p(\textbf{x})}{p(\textbf{x}) + q(\textbf{x})} \text{ and hence } \frac{P(z = 1 | \textbf{x})}{P(z = -1 | \textbf{x})} = \frac{p(\textbf{x})}{q(\textbf{x})}
$$
use a logarithm regression method, let $P(z = 1| \textbf{x}) = \frac1{1 + \exp(-h(\textbf{x}))}$, then
$$
β_i = \frac{1/(1 + \exp(-h(\textbf{x}_i)))}{\exp(-h(\textbf{x}_i))/(1 + \exp(-h(\textbf{x}_i)))} = \exp(h(\textbf{x}_i))
$$

suppose a training set $\{(\textbf{x}_1, y_1), ..., (\textbf{x}_n, y_n)\}$ and an unlabelled testing set $\{\textbf{u}_1, ..., \textbf{u}_m\}$, suppose $1 ≤ i ≤ n, \textbf{x}_i$ from source dist, and $\textbf{u}_i$ from target dist
algorithm:
1. generate a dual-classification training set $\{(\textbf{x}_1, -1), ..., (\textbf{x}_n, -1), (\textbf{u}_1, 1),..., (\textbf{u}_m, 1) \}$
2. use logarithm regression to train a classifier $h$
3. use $β_i = \exp(h(\textbf{x}_i))$ or $\beta_i = \min(h(\textbf{x}_i), c)$ to weigh training data
4. use $\beta_i$ to train $\{(\textbf{x}_1, y_1), ..., (\textbf{x}_n, y_n)\}$

#### 4.9.3.3 Label Shift Correction
suppose $q(y) ≠ p(y)$ but $q(\textbf{x} | y) = p(\textbf{x} | y)$
use correction formula
$$
∫∫l(f(\textbf{x}), y)p(\textbf{x} | y)p(y) d\textbf{x}dy = ∫∫l(f(\textbf{x}), y)q(\textbf{x} | y) q(y) \frac{p(y)}{q(y)}d\textbf{x}dy
$$
likewise
$$
β_i \stackrel{\text{def}}{=} \frac{p(y_i)}{q(y_i)}
$$

use mixture matrix $\textbf{C}_{k×k}, c_{ij}$ means, in validation set, true label is $j$ and the probability that the model predict $i$

average model output $μ(\hat{\textbf{y}}) \in \mathbb{R}^k, μ(\hat{y}_i)$ is the total predication score of $i$ in testing set.
$$
\textbf{C}p(\textbf{y}) = μ(\hat{\textbf{y}})
$$
