## 18.1. Introduction to Gaussian Process

$$
k_{RBF}(x, x') = \text{Cov}(f(x), f(x')) = a^2\exp(-\frac{1}{2\ell^2}||x - x'||^2)
$$

$$
\left[\begin{aligned}
f(x) \\ 
f(x_1) \\ 
\vdots \\ 
f(x_n) 
\end{aligned}\right] \sim \mathcal{N}\left(\mu, 
\left[
  \begin{aligned}
  k(x, x)   & k(x, x_1)   & ... & k(x, x_n) \\ 
  k(x_1, x) & k(x_1, x_1) & ... & k(x_1, x_n) \\ 
  ⋮          &    \quad ⋮       & ... & \quad ⋮ \\ 
  k(x_n, x) & k(x_n, x_1) & ... & k(x_n, x_n)
  \end{aligned}
\right]\right)
$$

$$
f(x) | f(x_1), ..., f(x_n) \sim \mathcal{N}(m, s^2)
$$

$$
m = k(x, x_{1:n})k(x_{1:n}, x_{1:n})^{-1} f(x_{1:n}) \\ 
s^2 = k(x, x) - k(x, x_{1:n})k(x_{1:n}, x_{1:n})^{-1}k(x, x_{1:n}) \\ 
$$

$$
\left[\begin{aligned}
f(x)
f(x_1)
\end{aligned}\right] ∼ \mathcal{N}(μ, 
\left[
\begin{aligned}
k(x, x) & k(x, x_1) \\ 
k(x_1, x) & k(x_1, x_1)
\end{aligned}
\right])
$$

$$
m = k(x, x_{1:3})k(x_{1:3}, x_{1:3})^{-1} f(x_{1:3}) \\ 
s^2 = k(x, x) - k(x, x_{1:3})k(x_{1:3}, x_{1:3})^{-1}k(x, x_{1:3})
$$
