## 6.3. Padding and Stride

### 6.3.1 Padding
padding with height $p_h$ and width $p_w$, new output size:
$$
(n_h - k_h + p_h + 1) × (n_w - k_w + p_w + 1)
$$

if $p_h = k_h - 1, p_w = k_w - 1$ then output and input share the same size

```python
import torch
from torch import nn

# define a Conv Layer
# init weight, and change with input, output
def comp_conv2d(conv2d, X):
  # (1, 1) means batch_size and channel are all 1
  X = X.reshape((1, 1) + X.shape)
  Y =  conv2d(X)
  # leave the first two dimension, batch_size and channel
  return Y.reshape(Y.shape[2:])

# each sides padding 1 row, so total 2 row and 2 column
conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1)
X = torch.rand(size=(8, 8))
comp_conv2d(conv2d, X).shape
```

```python
conv2d = nn.Conv2d(1, 1, kernel_size=(5, 3), padding=(2, 1))
comp_conv2d(conv2d, X).shape
```

### 6.3.2 Stride
when verticale stride is $s_hS and horizontal stride is $s_w$, output size is:
$$
⌊ (n_h - k_h + p_h + s_h) / s_h ⌋ × ⌊ (n_w - k_w + p_w + s_w) / s_w ⌋ 
$$

if $p_h = k_h - 1$ and $p_w = k_w - 1$, then $⌊(n_h + s_h - 1) / s_h⌋ × ⌊(n_w + s_w - 1) / s_w⌋$

if input size could be divided by stride, then output size is $(n_h / s_h) × (n_w / s_w)$

```python
conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1, stride=2)
comp_conv2d(conv2d, X).shape
```

```python
conv2d = nn.Conv2d(1,1,kernel_size=(3,5), padding=(0, 1), stride=(3, 4))
comp_conv2d(conv2d, X).shape
```

in practice, usually $p_h = p_w$ and $s_h = s_w$, thence denote them as $p, s$
