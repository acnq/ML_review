## 8.4. Recurrent Neural Networks (RNN)

$$
P(x_t | x_{t - 1}, ..., x_1) ≈ P(x_t | h_{t - 1}) \\ 
h_t = f(x_t, h_{t - 1})
$$

### 8.4.1 Neural Network without Hidden State

$$
\textbf{H} = ϕ(\textbf{XW}_{xh} + \textbf{b}_h) \\ 
\textbf{O} = \textbf{HW}_{hq} + \textbf{b}_q
$$

### 8.4.2 Recurrent Neural Network with Hidden State

$$
\textbf{H}_t = ϕ(\textbf{X}_t\textbf{W}_{xh} + \textbf{H}_{t-1}\textbf{W}_{hh} + \textbf{b}_h) \\ 
\textbf{O}_t = \textbf{H}_t\textbf{W}_{hq} + \textbf{b}_q
$$

```python
import torch
from d2l import torch as d2l

X, W_xh = torch.normal(0, 1, (3, 1)), torch.normal(0, 1, (1, 4))
H, W_hh = torch.normal(0, 1, (3, 4)), torch.normal(0, 1, (4, 4))
torch.matmul(X, W_xh) + torch.matmul(H, W_hh)
```

```python
torch.matmul(torch.cat((X, H), 1), torch.cat((W_xh, W_hh), 0))
```
