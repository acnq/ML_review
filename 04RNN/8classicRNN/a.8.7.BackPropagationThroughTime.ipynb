## 8.7. BackPropagation through Time (BPTT)

### 8.7.1 RNN gradient analysis

$$
\begin{aligned}
h_t &= f(x_t, h_{t-1}, w_h) \\ 
o_t &= g(h_t, w_o)
\end{aligned}
$$

$$
L(x_1, ..., x_T, y_1, ..., y_T, w_h, w_o) = \frac1T\sum_{t=1}^Tl(y_t, o_t)
$$

$$
\begin{aligned}
\frac{∂L}{∂ w_h}  &= \frac1T\sum_{t=1}^T\frac{∂l(y_t, o_t)}{∂ w_h} \\ 
                  &= \frac1T\sum_{t=1}^T\frac{∂l(y_t, o_t)}{∂ o_t} \frac{∂ g(h_t, w_o)}{∂ h_t} \frac{∂ h_t}{∂ w_h}
\end{aligned} 
$$

$$
\frac{∂ h_t}{\partial w_h} = \frac{∂ f(x_t, h_{t-1}, w_h)}{∂ w_h} + \frac{∂ f(x_t, h_{t - 1}, w_h)}{∂ h_{t-1}} \frac{\partial h_{t-1}}{∂ w_h}
$$

suppose three sequence, $a_0 = 0, a_t = b_t + c_ta_{t-1}$
$$
a_t = b_t + \sum_{i=1}^{t - 1}(\prod_{j=i+1}^t c_j)b_i
$$

$$
\begin{aligned}
a_t &= \frac{∂ h_t}{∂ w_h} \\ 
b_t &= \frac{\partial f(x_t, h_{t-1}, w_h)}{∂ w_h} \\ 
c_t &= \frac{∂ f(x_t, h_{t-1}, w_h)}{∂ h_{t-1}}
\end{aligned}
$$

$$
\frac{∂ h_t}{\partial w_h} = \frac{∂ f(x_t, h_{t-1}, w_h)}{∂ w_h} + \sum_{i=1}^{t-1}(\prod_{j=i+1}^t \frac{∂ f(x_j, h_{j-1}, w_h)}{∂ h_{j-1}})\frac{∂ f(x_i, h_{i-1}, w_h)}{∂ w_h}
$$

### 8.7.2 detail of BPTT

$$
\begin{aligned}
\textbf{h}_t &= \textbf{W}_{hx}\textbf{x}_t + \textbf{W}_{hh}\textbf{h}_{t-1} \\ 
\textbf{o}_t &= \textbf{W}_{qh}\textbf{h}_t \\ 
L &= \frac1T\sum_{t=1}^Tl(\textbf{o}_t, y_t)
\end{aligned}
$$

$$
\begin{aligned}
\frac{∂ L}{∂ \textbf{o}_t} &= \frac{\partial l(\textbf{o}_t, y_t)}{T ⋅ ∂ \textbf{o}_t} \in \mathbb{R}^q \\ 
\frac{∂ L}{∂ \textbf{W}_{qh}} &= \sum_{t=1}^T\text{prod}(\frac{∂ L}{∂ \textbf{o}_t}, \frac{∂ \textbf{o}_t}{∂ \textbf{W}_{qh}}) &= \sum_{t=1}^T \frac{∂ L}{∂ \textbf{o}_t}\textbf{h}_t^⊤ \\ 
\frac{∂ L}{∂ \textbf{h}_T} &= \text{prod}(\frac{\partial L}{∂ \textbf{o}_T}, \frac{∂ \textbf{o}_T}{∂ \textbf{h}_T}) &= \textbf{W}_{qh}^⊤ \frac{∂ L}{∂ \textbf{o}_T} \\ 
\frac{∂ L}{∂ \textbf{h}_t} &= \text{prod}(\frac{∂ L}{∂ \textbf{h}_{t+1}}, \frac{∂ \textbf{h}_{t+1}}{∂ \textbf{h}_t}) + \text{prod}(\frac{∂ L}{∂ \textbf{o}_t}, \frac{∂ \textbf{o}_t}{∂ \textbf{h}_t}) &= \textbf{W}_{hh}^⊤\frac{∂ L}{∂ \textbf{h}_{t+1}} + \textbf{W}_{qh}^⊤\frac{∂ L}{∂ \textbf{o}_t}
\end{aligned} \\ 
\frac{∂ L}{∂ \textbf{h}_t} = \sum_{i=t}^T(\textbf{W}_{hh}^\top)^{T-i}\textbf{W}_{qh}^⊤\frac{∂ L}{∂ \textbf{o}_{T+t-i}}
$$

$$
\begin{aligned}
\frac{∂ L}{∂ \textbf{W}_{hx}} &= \sum_{t=1}^T\text{prod}(\frac{∂ L}{∂ \textbf{h}_t}, \frac{∂ \textbf{h}_t}{∂ \textbf{W}_{hx}}) &= \sum_{t=1}^T\frac{∂ L}{∂ \textbf{h}_t}\textbf{x}_t^⊤, \\ 
\frac{∂ L}{∂ \textbf{W}_{hh}} &= \sum_{t=1}^T\text{prod}(\frac{∂ L}{∂ \textbf{h}_t}, \frac{∂ \textbf{h}_t}{∂ \textbf{W}_{hh}}) &= \sum_{t=1}^T\frac{∂ L}{∂ \textbf{h}_t}\textbf{h}_{t-1}^⊤,
\end{aligned}
$$
