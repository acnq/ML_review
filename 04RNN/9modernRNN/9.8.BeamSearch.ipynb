### 9.8.3 Beam Search

$$
\begin{aligned}
P(A, y_2 | \textbf{c}) &= P(A | \textbf{c}) P(y_2 | A, \textbf{c}) \\ 
P(C, y_2 | \textbf{c}) &= P(C | \textbf{c}) P(y_2 | C, \textbf{c}) \\ 
\end{aligned}
$$

$$
\begin{aligned}
P(A, B, y_3 | \textbf{c}) &= P(A, B | \textbf{c}) P(y_3 | A, B, \textbf{c}) \\ 
P(C, E, y_3 | \textbf{c}) &= P(C, E | \textbf{c}) P(y_3 | C, E, \textbf{c}) \\ 
\end{aligned}
$$

$$
\frac1{L^α}\log P(y_1, ..., y_L | \textbf{c}) = \frac1{L^α}\sum_{t'=1}^L\log P(y_{t'} | y_1, ..., y_{t'-1}, \textbf{c})
$$
