### 14.1.2. Self-Supervised word2vec

### 14.1.3 Skip-Gram

$$
P(\text{"the"}, \text{"man"}, \text{"this"}, \text{"son"} | \text{"loves"})
$$

assume conditionally independent
$$
P(\text{"the"} | \text{"loves"}) ⋅ P(\text{"man"} | \text{"loves"}) ⋅ P(\text{"his"} | \text{"loves"}) ⋅ P(\text{"son"} | \text{"loves"})
$$

skip-gram model
$$
P(w_o | w_c) = \frac{\exp(\textbf{u}_o^⊤\textbf{v}_c)}{\sum_{i \in \mathcal{V}} \exp(\textbf{u}_i^⊤\textbf{v}_c)}
$$

$$
\prod_{t=1}^T\prod_{-m≤j≤m, j≠0} P(w^{(t+j)} | w^{(t)})
$$

#### 14.1.3.1 Training

$$
-\sum_{t=1}^T\sum_{-m ≤j ≤m, j≠0} \log P(w^{(t + j)} | w^{(t)})
$$

$$
\log P(w_o | w_c) = \textbf{u}_o^⊤\textbf{v}_c - \log(\sum_{i\in \mathcal{V}}\exp(\textbf{u}_i⊤\textbf{v}_c))
$$

$$
\begin{aligned}
\frac{∂ \log P(w_o | w_c)}{∂ \textbf{v}_c} &= \textbf{u}_o - \frac{\sum_{j \in \mathcal{V}} \exp(\textbf{u}_j⊤\textbf{v}_c)\textbf{u}_j}{\sum_{i∈\mathcal{V}}\exp(\textbf{u}_i^⊤\textbf{v}_c)} \\ 
&= \textbf{u}_o - \sum_{j \in \mathcal{V}} (\frac{ \exp(\textbf{u}_j⊤\textbf{v}_c)}{\sum_{i∈\mathcal{V}}\exp(\textbf{u}_i^⊤\textbf{v}_c)}) \textbf{u}_j \\ 
&= \textbf{u}_o - \sum_{j∈\mathcal{V}}P(w_j | w_c) \textbf{u}_j
\end{aligned}
$$

#### 14.1.4 Continuous Bag of Word (CBOW)

$$
P(\text{"loves"} | \text{"the"}, \text{"man"}, \text{"his"}, \text{"son"})
$$

$$
P(w_c | w_{o_1}, ..., w_{o_{2m}}) = \frac{\exp(\frac{1}{2m} \textbf{u}_c^⊤(\textbf{v}_{o_1} + ... + \textbf{v}_{o_{2m}}))}{\sum_{i\in \mathcal{V}}\exp({\frac{1}{2m}\textbf{u}_i^⊤(\textbf{v}_{o_1} + ... + \textbf{v}_{o_{2m}})})}
$$

$$
P(w_c | \mathcal{W}_o) = \frac{\exp(\textbf{u}_c^⊤\bar{\textbf{v}}_o)}{\sum_{i∈\mathcal{V}}\exp(\textbf{u}_i^⊤)}
$$

$$
\prod_{t=1}^TP(w^{(t)} | w^{t-m}, ...,w^{t-1}, w^{t+1},...,w^{t+m})
$$

#### 14.1.4.1 Training

$$
-\sum_{t=1}^T\log P(w^{(t)} | w^{(t-m)}, ..., w^{(t-1)}, w^{(t+1)}, ..., w^{(t+m)})
$$

$$
\log P(w_c | \mathcal{W}_o) = \textbf{u}_c^⊤ \bar{\textbf{v}}_o - \log(\sum_{i \in \mathcal{V}}\exp(\textbf{u}_i^⊤\bar{\textbf{v}}_o))
$$

$$
\frac{∂ \log P(w_c | \mathcal{W}_o)}{∂ \textbf{v}_{o_i}} = \frac1{2m}(\textbf{u}_c - \sum_{j∈\mathcal{V}}\frac{\exp(\textbf{u}^⊤_j\bar{\textbf{v}}_o)\textbf{u}_j}{\sum_{i∈\mathcal{V}} \exp(\textbf{u}_i^⊤\bar{\textbf{v}}_o)}) = \frac{1}{2m}(\textbf{u}_c - \sum_{j \in \mathcal{V}}P(w_j | \mathcal{W}_o)\textbf{u}_j)
