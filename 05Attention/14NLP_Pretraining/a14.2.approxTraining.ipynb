## 14.2. Approximation Training

### 14.2.1 Negative Sampling

$D:$ whether $w_o$ comes from context window
$$
P(D = 1 | w_c, w_o) = σ(\textbf{u}_o^⊤\textbf{v}_c) \\ 
σ(x) = \frac{1}{1 + \exp(-x)} \\ 
\prod_{t=1}^T∏_{-m≤j≤m,j≠0} P(D = 1 | w^{(t)}, w^{(t + j)}) \\ 
$$

$$
P(w^{(t + j)} | w^{(t)}) = P(D = 1 | w^{(t)}, w^{(t + j)}) \prod_{k=1, w_k \sim P(w)}^K P(D = 0 | w^{(t)}, w_k)
$$

$$
\begin{aligned}
-\log P(w^{(t + j)} | w^{(t)}) &= -\log P(D = 1 | w^{(t)}, w^{(t + j)}) - \sum_{k=1, w_k \sim P(w)}^K \log P(D = 0 | w^{(t)}, w_k) \\ 
&= - \log σ(\textbf{u}_{i_{t+j}}^⊤ v_{i_t}) - \sum_{k=1, w_k \sim P(w)}^K \log(1 - σ(\textbf{u}_{h_k}^⊤ v_{i_t})) \\ 
&= - \log σ(\textbf{u}_{i_{t + j}}^⊤ \textbf{v}_{i_t}) - \sum_{k=1, w_k \sim P(w)}^K \log σ(-\textbf{u}^⊤_{h_k}\textbf{v}_{i_t})
\end{aligned} 
$$
