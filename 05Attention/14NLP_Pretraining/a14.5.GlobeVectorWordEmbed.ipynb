## 14.5. Word Embedded of Globle Vector (GloVe)

### 14.5.1 Skip-Gram with Globle Corpus Statistics

$$
q_{ij} = \frac{\exp(\textbf{u}_j^\top\textbf{v}_i)}{\sum_{k\in\mathcal{V}}\exp(\textbf{u}_k^\top\textbf{v}_i)}
$$

$$
-\sum_{i\in\mathcal{V}}\sum_{j\in\mathcal{V}}x_{ij}\log q_{ij}
$$

$$
-\sum_{i\in\mathcal{V}}x_i\sum_{j\in\mathcal{V}}p_{ij}\log q_{ij}
$$

### 14.5.2 GloVe Model

$$
\sum_{i∈\mathcal{V}}\sum_{j∈\mathcal{V}}h(x_{ij})(\textbf{u}_j^⊤\textbf{v}_i + b_i + c_j - \log x_{ij})^2
$$

### 14.5.3 Understand GloVe Model as Conditional Probability

$$
f(\textbf{u}_j, \textbf{u}_k, \textbf{v}_i) ≈ \frac{p_{ij}}{p_{ik}}
$$

$$
f(\textbf{u}_j, \textbf{u}_k, \textbf{v}_i) = \frac{\exp(\textbf{u}_j^⊤\textbf{v}_i)}{\exp(\textbf{u}_k^⊤\textbf{v}_i)} ≈ \frac{p_{ij}}{p_{ik}}
$$

$$
\textbf{u}_j^⊤\textbf{v}_i + b_i + c_j ≈ \log x_{ij}
$$
