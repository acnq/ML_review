### 6.1.2 Limitation of MLP
new input: 2D fig, represented by a matrix $\textbf{X}$, hidden layer also a matrix $\textbf{H}$

$[\textbf{X}]_{i,j}$ and $[\textbf{H}]_{i,j}$ denote the pixel at position $(i, j)$

now the weights need to be a 4-degree tensor $\mathsf{W}$, and bias also becomes a matrix $\textbf{U}$

formalize FLC, 
$$ 
\begin{aligned}
[\textbf{H}]_{i,j}  &= [\textbf{U}]_{i,j} + \sum_k\sum_l[\mathsf{W}]_{i,j,k,l}[\textbf{X}]_{k,l} \\ 
&= [\textbf{U}]_{i,j} + \sum_a\sum_b[\mathsf{V}]_{i,j,a,b}[\textbf{X}]_{i+a,j+b} \\ 
\end{aligned}
$$
the second line is just reindex, i.e. $k = i + a, l = j + b => [\mathsf{V}]_{i,j,a,b} = [\mathsf{W}]_{i,j,i+a,j+b}$

#### 6.1.2.1 Translation Invariance
$\mathsf{V}, \textbf{U}$ is not dependent to $(i, j)$, i.e. $[\mathsf{V}]_{i,j,a,b} = [\textbf{V}]_{a,b}, [\textbf{U}]_{i,j} = u$
thus:
$$
[\textbf{H}]_{i,j} = u + \sum_a\sum_b[\textbf{V}]_{a, b}[\textbf{X}]_{i+a, j+b}
$$

#### 6.1.2.2 Locality
don't collect information for $[\textbf{H}]_{i,j}$ from places too far from $(i,j)$， so set $Δ$, let $[\textbf{V}]_{a,b}=0$ where $|a| > Δ, |b| > Δ$

$$
[\textbf{H}]_{i,j} = u + \sum_{a=-Δ}^\Delta \sum_{b=-Δ}^\Delta[\textbf{V}]_{a, b}[\textbf{X}]_{i+a, j+b}
$$

$\textbf{V}$ means a 'convolutional kernel' or 'filter', and $\textbf{H}$ means a 'convolutional layer'

### 6.1.3 Convolution

definition:
$$
(f * g)(\textbf{x}) = ∫ f(\textbf{z})g(\textbf{x} - \textbf{z}) d\textbf{z}
$$
or
$$
(f * g)(i) = \sum_a f(a)g(i-a)
$$
2d tensor
$$
(f * g)(i,j) = \sum_a\sum_b f(a, b)g(i-a, j-b)
$$
